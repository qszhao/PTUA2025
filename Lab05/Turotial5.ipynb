```python
##Q18
df["day_of_week"] = pd.to_datetime(df.index).dayofweek  # (0=Monday, 6=Sunday)

df_weekday = df[df["day_of_week"] < 5]  # 选择周一到周五 (0-4)
df_weekend = df[df["day_of_week"] >= 5]  # 选择周六、周日 (5-6)

# 显示前几行，检查数据
print("Weekday DataFrame (First 5 rows):")
print(df_weekday.head())

print("\nWeekend DataFrame (First 5 rows):")
print(df_weekend.head())

# 保存为 CSV 文件
df_weekday.to_csv("weekday_data.csv")
df_weekend.to_csv("weekend_data.csv")
print("Weekday and Weekend DataFrames saved as 'weekday_data.csv' and 'weekend_data.csv'")
```

```python
##Q19
observations_per_day = df["day_of_week"].value_counts().sort_index()
plt.figure(figsize=(8, 5))
plt.bar(observations_per_day.index, observations_per_day.values)
plt.xlabel("Day of the Week (0=Monday, 6=Sunday)")
plt.ylabel("Number of Observations")
plt.title("Number of Observations per Day of the Week")
plt.xticks(ticks=range(7), labels=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
plt.show()
```
